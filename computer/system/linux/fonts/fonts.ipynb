字体分类 字体名称 特点简介 安装命令 (Ubuntu)

中文字体 思源宋体 (Source Han Serif / Noto Serif CJK) Adobe 与 Google 联合开发，Serif 衬线字体，印刷质感强，非常适合学术论文正文 sudo apt install fonts-noto-cjk

思源黑体 (Source Han Sans / Noto Sans CJK) Adobe 与 Google 联合开发，Sans-Serif 无衬线字体，现代简洁，适合标题和图表注释 sudo apt install fonts-noto-cjk (通常一并安装)

文泉驿正黑 (WenQuanYi Zen Hei) 开源中文字体，清晰易读，对简体中文支持良好 sudo apt install fonts-wqy-zenhei

文泉驿微米黑 (WenQuanYi Micro Hei) 开源中文字体，屏幕显示效果佳，适合长时间阅读 sudo apt install fonts-wqy-microhei

英文字体 DejaVu Fonts 高质量开源字体家族，覆盖字符范围极广，是许多Linux系统的默认字体 sudo apt install fonts-dejavu

Liberation Fonts 旨在替代 Windows 常用字体（如 Times New Roman, Arial），兼容性好 sudo apt install fonts-liberation

Noto Sans Google 开发，“无豆腐”字体，旨在消除所有空白框（□），支持几乎所有语言 sudo apt install fonts-noto

Ubuntu Font Family Ubuntu 系统官方字体，设计现代，屏幕可读性优化 通常已预装，或 sudo apt install fonts-ubuntu

Source Sans Pro Adobe 开发的首款开源无衬线字体，设计优雅，在学术界也很受欢迎 可能需要手动下载安装

```python
# 首先更新软件包列表
sudo apt update

# 安装你选中的字体包，例如同时安装思源宋体、文泉驿正黑和 Liberation 字体：
sudo apt install fonts-noto-cjk fonts-wqy-zenhei fonts-liberation

# 安装后刷新字体缓存，让系统识别新字体
sudo fc-cache -fv
```
