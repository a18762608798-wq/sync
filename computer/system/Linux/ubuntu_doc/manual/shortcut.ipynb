## 2 desktop

系统创建的一般在：/usr/share/applications，可以查一下默认名字。

### 2.1 开机自启

自启目录：ls ~/.config/autostart （没有自创：mkdir -p ~/.config/autostart）



```python
# 查看自启目录
ls ~/.config/autostart
mkdir -p ~/.config/autostart # （倘若没有自创）
# 创建desktop文件，并写入
nano ~/.config/autostart/fcitx5.desktop
```

```python
[Desktop Entry]
Type=Application
Name=Fcitx5
Comment=Start Fcitx5 input method
Exec=fcitx5
Categories=System;
X-GNOME-Autostart-enabled=true
```
