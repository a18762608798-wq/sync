## 1.1 分区

根据Linux文件系统层次标准（FHS），根目录下除了`/opt`和`/usr`外，还有以下主要目录：

## **根目录完整列表**
- `/bin` - 基本可执行程序 
- `/boot` - 系统启动文件 
- `/dev` - 设备文件
- `/etc` - 系统配置文件 
- `/home` - 普通用户主目录 
- `/lib`/`/lib64` - 库文件
- `/media` - 媒体设备挂载点 
- `/mnt` - 临时挂载目录 
- `/proc` - 进程和内核信息
- `/root` - root用户主目录（不是根目录/）
- `/sbin` - 系统管理命令 
- `/srv` - 服务数据
- `/sys` - 系统信息
- `/tmp` - 临时文件 
- `/var` - 可变数据（日志、缓存等）

## **最适合放到机械盘的目录**

### **✅ 强烈推荐（完美适合机械盘）：**
1. **`/home`** - 用户个人文件、文档、下载等，通常占用大量空间且访问频率相对较低 
2. **`/srv`** - 服务数据目录，适合存放大型服务数据（*但是其实很小，不必单独放在机械盘*）
3. **`/media`和`/mnt`** - 专门设计用于挂载外部存储设备 （*同样很很小，毫无必要性。*）

### **✅ 推荐（适合机械盘）：**
4. **`/opt`** - 您之前提到的第三方应用程序，完全适合机械盘
5. **`/var`** - 日志文件、邮件、缓存等，虽然部分需要性能，但大部分数据可以放机械盘 

### **❌ 不适合（必须放在SSD）：**
- `/bin`, `/sbin`, `/lib` - 系统启动必需，需要高速访问 
- `/boot` - 启动文件，影响开机速度
- `/etc` - 配置文件，频繁读取
- `/usr` - 程序文件，影响应用启动速度
- `/tmp` - 临时文件，需要高速读写 


总之划分给机械盘：
* home
* opt
* var

# 2 package management

## 2.1 install

建议不用用本地.deb 而是官方库，会更新。

除了apt，目前还用到flatpak/snap. 这些偏向桌面deb, 可以**自动更新**，但是apt不可代替。

建议开摆全部用apt.

```julia
# 更新系统软件包
sudo apt update  # check
sudo apt upgrade  # upgrade all the packages

# install package（库里面有的可以直接用默认名，卸载是remove）
sudo apt install /下载.../software.deb # 本地下载，会自动识别并配置
```

## 2.2 activate

对于deb总有可执行文件在$PATH. 对于非默认路径可以在文档 ‘~/.profile’ 添加：

export PATH=$PATH:/path/to/your/dir

随后激活：

source ~/.profile

### 2.2.1 apt

```julia
# 寻找可执行文件
apt list --installed | grep -i python # 彻底安装过的应用忘了叫什么，关键词硬找吧。(-i忽略大小写)
ls /usr/bin/ | grep qq # 找完整可执行文件名字比较好，如果在这里。
which qq

# 软件包信息
apt show <软件包名> # 需要准全包名，看info(这有什么用)
dpkg -L <软件包名> # get relative path

# 执行
qq # 直接执行可执行文件
qq & # 序在后台运行
```
