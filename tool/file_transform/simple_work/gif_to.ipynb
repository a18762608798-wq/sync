# gif to mp4

````python
import os
import re
import subprocess

def natural_sort_key(s):
    """
    辅助函数，用于按自然语言顺序对字符串排序。
    """
    return [int(text) if text.isdigit() else text.lower() for text in re.split('([0-9]+)', s)]

def find_gif_files(main_folder):
    """
    遍历主文件夹及所有子文件夹，查找 GIF 文件。
    返回按自然语言排序的文件路径列表。
    """
    gif_files = []
    for root, dirs, files in os.walk(main_folder):
        for file in files:
            if file.lower().endswith('.gif'):
                gif_files.append(os.path.join(root, file))
    gif_files.sort(key=natural_sort_key)  # 使用自然排序
    return gif_files

main_folder = input("请输入主文件夹路径: ").strip()

if not os.path.isdir(main_folder):
    print("提供的路径无效，请检查后重试。")
else:
    gif_files = find_gif_files(main_folder)
    if not gif_files:
        print("未找到任何 GIF 文件，请检查文件夹内容。")
    else:
        print(f"找到 {len(gif_files)} 个 GIF 文件。")

        filelist_path = os.path.join(main_folder, "filelist.txt")
        with open(filelist_path, "w", encoding="utf-8") as f:
            for gif in gif_files:
                escaped_path = gif.replace("'", r"'\''")  # 处理路径中的单引号
                f.write(f"file '{escaped_path}'\n")
        print(f"已生成文件列表: {filelist_path}")

        output_path = os.path.join(main_folder, "output.mp4")
        ffmpeg_command = [
            "ffmpeg",
            "-f", "concat",
            "-safe", "0",
            "-i", filelist_path,
            "-vf", "fps=25,scale=iw:-2,format=yuv420p",  # 修改点
            "-c:v", "libx264",
            "-pix_fmt", "yuv420p",
            output_path
        ]

        try:
            print("正在合并 GIF 文件...")
            subprocess.run(ffmpeg_command, check=True, text=True)
            print(f"合并完成，输出文件位于: {output_path}")
        except subprocess.CalledProcessError as e:
            print("FFmpeg 执行失败，错误信息如下：")
            print(e.stderr)
        
        try:
            os.remove(filelist_path)
            print(f"已删除临时文件: {filelist_path}")
        except Exception as e:
            print(f"无法删除临时文件 {filelist_path}：{e}")
````
